```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("istherecorrelation.csv", sep=";", decimal=",", encoding="utf-8-sig")

x = df["WO [x1000]"]
y = df["NL Beer consumption [x1000 hectoliter]"]

plt.scatter(x, y)

a, b = np.polyfit(x, y, 1)
plt.plot(x, a * x + b, color="red")

for i in range(len(df)):
    plt.annotate(df["Year"][i], (x[i], y[i]))

plt.xlabel("WO-students [x1000]")
plt.ylabel("Bier Consumption NL [x1000 hectoliter]")
plt.title("WO-students vs. beer consumption")
plt.savefig("plot.png", dpi=300)
plt.show()
```
